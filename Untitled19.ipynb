```python
#count the occurences of each word in a line of text
text = input("Enter a line of text:")
words = text.split()
count ={}
for word in words:
    if word in count:
        count[word] +=1
    else:
        count[word]=1
print("word occurences:")
for word in count:
    print(word,":",count[word])
```

```python
#prompt the user for a list of integers.for all values greater than 100,store "ove" instead
numbers = list(map(int,input("Enter integers:").split()))
for i in range(len(numbers)):
    if numbers[i]>100:
        numbers[i]="over"
print(numbers)
```

```python
#store a list of names.count the occurence of "a" wihin list
names=input("Enter first names separated by spaces:").split()
count=0
for n in names:
    count+= n.lower().count('a')
    print("Number of occurences:",count)
```

```python
#find gcd of 2 numbers
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
while b !=0:
    a,b=b,a%b
    print("GCD=",a)
```

```python
# from a list of integers,create a list removing even numbers
numbers = list(map(int,input("Enter integers:").split()))
result =[]
for num in numbers:
     if num %2 !=0:
         result.append(num)
print("List after removing even numbers:",result)
```

```python
#program to find factorial of a number
n =int(input("Enter a number:"))
fact = 1
for i in range(1,n + 1):
    fact = fact * i
    print("factorial =",fact)
```

```python
#generate a list of four digit numbers in a given range with all their digits even and the number is a perfect square
import math
start=int(input("Enter starting range:"))
end=int(input("Enter ending range:"))
result=[]
for num in range(start,end+1):
    if 1000<=num<=9999:
        if int(math.sqrt(num))**2==num:
            if all(int(digit)%2==0 for digit in str(num)):
                result.append(num)
print("result:",result)
                
```

```python
#character frequency
text=input("Enter a string:")
count={}
for char in text:
    count[char]=count.get(char,0)+1
print("character frequency:")
for char,frequency in count.items():
    print(char,"",frequency)
```

```python
#compare two lists
l1=list(map(int,input("Enter first list:").split()))
l2=list(map(int,input("Enter second list:").split()))
#check same length
if len(l1)==len(l2):
    print("Both lists have the same length")
else:
       print("Both lists donot have the same length")
#check the same sum
if sum(l1)==sum(l2):
    print("Both LISTS HAVE THE SAME SUM")
else:
    print("Both LISTS donot HAVE THE SAME SUM")
#check the same value
if set(l1)&set(l2):
    print("both list have common value")
else:
        print("No common values")
```
