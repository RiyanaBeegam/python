```python
#change a given string to new string
a=input("Enter a string:")
b=a[-1]+a[1:-1]+a[0]
print("new string:",b)
```

```python
#to get a string made of the 1st and last characters
a=input("Enter a string:")
b=a[1:2]+a[-2:]
print("New string:",b)
```

```python
#swapping
s1=input("Enter a string:")
s2=input("Enter a string:")
a=s2[0]+s1[1:]
b=s1[0]+s2[1:]
print("new string:",a,b)
```

```python
#replacing  character
a=input("Enter a string:")
b=a[0]+a[1:].replace(a[0],"$")
print("new string:",b)
```
