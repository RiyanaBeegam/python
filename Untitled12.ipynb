```python
#accept list of words and return the length of longest word
def longestword():
    words=input("Enter words(separated by space):)").split()
    longest=max(words,key=len)
    print("Length of the longest word is :",len(longest))
longestword()    
```

```python
#construting pattern using nested loop
def pattern():
    n=5
    for i in range(1,n+1):
        for j in range(i):
            print("*",end=" ")
        print()
    for i in range(n-1,0,-1):
        for j in range(i):
            print("*",end=" ")
        print()
pattern()      
```

```python
#generate all factors of a number
def factors():
    num=int(input("Enter a number:"))
    print("factors of","are:")
    for i in range(1,num+1):
        if num %i==0:
            print(i)
factors()        
```

```python
#lambda function to find area of square,rectangle and triangle
square=lambda side:side **2
rectangle=lambda length,breadth:length*breadth
triangle=lambda base,height:0.5*base*height
s=float(input("enter the side of square:"))
print("Area of square:",square(s))
l=float(input("enter the length of rectangle:"))
b=float(input("enter the breadth of rectangle:"))
print("Area of rectangle:",rectangle(l,b))
base=float(input("enter the base of triangle:"))
height=float(input("enter the base of triangle:"))
print("Area of triangle:",triangle(base,height))
```
