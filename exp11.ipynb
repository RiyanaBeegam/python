```python
#generate  posite numbers from given list.
numbers = [-5, 10, -3, 7,0,4]
positive = [x for x in numbers if x > 0]
print("Positive numbers:",positive)
```

```python
# square numbers
n = int(input("Enter the number of elements:"))
numbers = [int(input("Enter a number:")) for i in range(n)]
squares = [x*x for x in numbers]
print("Squares:",squares)
```

```python
# list of vowels
word = input("Enter a word:")
vowels = [ch for ch in word if ch.lower() in 'aeiou']
print("Vowels:",vowels)
```

```python
# Ordinal values
word = input("Enter a word: ")
ordinal = [ord(ch) for ch in word]
print("Ordinal values:",ordinal)
```
