```python
#Write a Python program to read a file line by line and store it into a list.
# Accept values from the user
n = int(input("Enter the number of values: "))

with open("sample.txt", "w") as file:
    for i in range(n):
        value = input("Enter a value: ")
        file.write(value + "\n")

# Read the file line by line and store it in a list
lines = []

with open("sample.txt", "r") as file:
    for line in file:
        lines.append(line.strip())

# Display the list
print("List of values:", lines)
```

```python
#Python program to copy odd lines of one file to other
# Create and write content into the source file
with open("file1.txt", "w") as f:
    n = int(input("Enter the number of lines: "))
    for i in range(n):
        line = input("Enter a line: ")
        f.write(line + "\n")

# Copy odd lines into another file
with open("file1.txt", "r") as f1:
    with open("file2.txt", "w") as f2:
        for i, line in enumerate(f1, start=1):
            if i % 2 != 0:
                f2.write(line)

# Display the contents of the second file
with open("file2.txt", "r") as f2:
    print("\nOdd lines copied to file2.txt:")
    print(f2.read())
```

```python
#Write a Python program to read each row from a given csv file and print a list of strings.
import csv

# Create a CSV file and accept values from the user
n = int(input("Enter the number of rows: "))

with open("sample.csv", "w", newline="") as file:
    writer = csv.writer(file)

    for i in range(n):
        name = input("Enter name: ")
        age = input("Enter age: ")
        place = input("Enter place: ")

        writer.writerow([name, age, place])

# Read each row and store it as a string in a list
data = []

with open("sample.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        data.append(','.join(row))

# Display the list
print("\nList of strings:")
print(data)
```

```python
#Write a Python program to write a Python dictionary to a csv file. After writing the CSV file read the CSV file and display the content
import csv

# Create a dictionary
data = [
    {'Name': 'Anu', 'Age': 20, 'Place': 'Kochi'},
    {'Name': 'Amal', 'Age': 21, 'Place': 'Kollam'},
    {'Name': 'Arun', 'Age': 22, 'Place': 'Trivandrum'}
]

# Write dictionary into CSV file
with open("student.csv", "w", newline="") as file:
    fieldnames = ['Name', 'Age', 'Place']
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(data)

# Read the CSV file and display its contents
with open("student.csv", "r") as file:
    reader = csv.reader(file)

    print("Contents of CSV file:")
    for row in reader:
        print(row)
```
