```python
#to sort dictionary in ascending and descending order
roll=input("Enter 5 roll numbers(space seperated)").split(",")
name=input("Enter 5 student names(space seperated)").split(",")
d=dict(zip(roll,name))
print("Dictionary",d)
print("Dictionary in ascending order",dict(sorted(d.items())))
print("Dictionary in descending order",dict(sorted(d.items(),reverse=True)))
```

```python
#accept two dictionaries and merge and display item
d1=dict(zip(input("Enter 3 roll numbers").split(","),input("Enter 3 students names").split(",")))
d2=dict(zip(input("Enter 2 roll numbers").split(","),input("Enter 2 students names").split(",")))
print("Dictionary1:",d1)
print("Dictionary2:",d2)
d1.update(d2)
print("Merged Dictionary:",d1)
```
