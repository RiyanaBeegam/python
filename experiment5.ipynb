```python
#to get the third element and third element from last of a tuple
t=(10,20,30,70,50,38)
print("3rd element:", t[2])
print("3rd element from last:", t[-3])
```

```python
#program which accepts a sequence of comma seprated numbers from user and generate a tuple with those numbers and find average of all items

t=input("enter elements (comma seperated):")
u=tuple(float(num.strip()) for num in t.split(','))
average=sum(u)/len(u)
print("Average of tuple:",average)
```
