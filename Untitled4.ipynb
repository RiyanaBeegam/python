```python
#list of colors from comma-separated color names entered by user,display first and last colors
colors=input("enter colors separated by comma:")
color_list=colors.split(",")
print("first color:",color_list[0])
print("second color:",color_list[-1])
```

```python
#python program to accept a filename from the user , print etension of that
filename=input("enter file name:")
extension=filename.split(".")[-1]
print("extension:",extension)
```

```python
#program which accepts a sequence of comma separated numbers from user and generate a list with those numbers and find sum of all items
numbers=input("enter comma separated numbers:")
num_list=[int(x) for x in numbers.split(",")]
print("list:",num_list)
print("sum:",sum(num_list))
```
