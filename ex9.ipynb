```python
 #program to add "ing" to the end of a given string.if string already end with "ing",add"ly" instead.
s=input("enter a string:")
if s.endswith("ing"):
    s=s+"ly"
else: 
    s=s+"ing"
print("result:",s)
```

```python
#check whether the string palindrome or not
s=input("enter a string:")
if s== s[::-1]:
    print("palindrome")
else:
    print("not a palindrome")
```
